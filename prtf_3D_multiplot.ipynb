```python
import numpy as np
import seaborn as sns
sns.set_theme()

from helper_functions import (write_text)

import matplotlib.pyplot as plt
import h5py
import glob

saveFigs = False
extension = 'prot_only' # 'prot_only' or 'prot_wat'
version = 'v_1' # 'v_1' or 'v_2' or more if more versions performed
```

```python
# 100k patterns
filesPRTF = np.sort(np.array(glob.glob(f'figures/*_100k_{extension}_*_{version}_PRTF.h5')))
names_PRTF = []
combs_PRTF = []
res_PRTF = []

i = 0
for file in filesPRTF:
    f_name = file.split(sep='/')[1].split(sep='.h5')[0][:-5]
    with h5py.File(file) as f:
        combs_PRTF.append(f['prtf_r_smooth'][:])
        res_PRTF.append(f['fp_res_inv_nm'][:])
        if 'res_r_smooth_nm' in f:
            res_r_smooth_nm = f['res_r_smooth_nm'][()]
            write_text(f'Resolution for [{i}] {f_name}: {res_r_smooth_nm} nm\n')
        else:
            write_text(f'Resolution for [{i}] {f_name}: - nm\n')
    i += 1
    names_PRTF.append(f_name)

names_PRTF = np.array(names_PRTF)
combs_PRTF = np.array(combs_PRTF)
res_PRTF = np.array(res_PRTF)
max_points = res_PRTF.shape[1]

res_ax = res_PRTF[0]

plt.figure(dpi=140)
plt.axhline(y=1/np.exp(1), xmin=0.0, xmax=res_ax[-1]*10, c='k', linestyle='--', linewidth=1.0, label='_nolegend_')
plt.axvline(x=1/0.6396483204611855, ymin=0 , ymax=1,c='k', linestyle='--', linewidth=1.0, alpha=0.4, label='_nolegend_')

for p in range(len(filesPRTF)):
    plt.plot(res_ax, combs_PRTF[p], linestyle='-')

plt.xlim([0, res_PRTF[-1][-1]])
plt.xlabel('|q| $(nm^{-1})$', weight='bold')
plt.ylabel('PRTF', weight='bold')
plt.legend(names_PRTF, frameon=True, prop=dict(size=5.0, weight='bold'), loc=1);
if saveFigs:
        plt.savefig(f'prtf_{extension}_100k_all.pdf', transparent=False, bbox_inches='tight',dpi=200);
```

```python
# 1M patterns
filesPRTF_1M = np.sort(np.array(glob.glob(f'figures/*_1M_{extension}_*_{version}_PRTF.h5')))
names_PRTF_1M = []
combs_PRTF_1M = []
res_PRTF_1M = []

i = 0
for file in filesPRTF_1M:
    f_name_1M = file.split(sep='/')[1].split(sep='.h5')[0][:-5]
    with h5py.File(file) as f_1M:
        combs_PRTF_1M.append(f_1M['prtf_r_smooth'][:])
        res_PRTF_1M.append(f_1M['fp_res_inv_nm'][:])
        if 'res_r_smooth_nm' in f_1M:
            res_r_smooth_nm_1M= f_1M['res_r_smooth_nm'][()]
            write_text(f'Resolution for [{i}] {f_name_1M}: {res_r_smooth_nm_1M} nm\n')
        else:
            write_text(f'Resolution for [{i}] {f_name_1M}: - nm\n')
    i += 1
    names_PRTF_1M.append(f_name_1M)

names_PRTF_1M = np.array(names_PRTF_1M)
combs_PRTF_1M = np.array(combs_PRTF_1M)
res_PRTF_1M = np.array(res_PRTF_1M)
max_points_1M = res_PRTF_1M.shape[1]

res_ax_1M = res_PRTF_1M[0]

plt.figure(dpi=140)
plt.axhline(y=1/np.exp(1), xmin=0.0, xmax=res_ax[-1]*10, c='k', linestyle='--', linewidth=1.0, label='_nolegend_')
plt.axvline(x=1/0.6396483204611855, ymin=0 , ymax=1,c='k', linestyle='--', linewidth=1.0, alpha=0.4, label='_nolegend_')

for p in range(len(filesPRTF_1M)):
    plt.plot(res_ax_1M, combs_PRTF_1M[p], linestyle='-')

plt.xlim([0, res_PRTF_1M[-1][-1]])
plt.xlabel('|q| $(nm^{-1})$', weight='bold')
plt.ylabel('PRTF', weight='bold')
plt.legend(names_PRTF_1M, frameon=True, prop=dict(size=5.0, weight='bold'), loc=1);
if saveFigs:
        plt.savefig(f'prtf_{extension}_1M_all.pdf', transparent=False, bbox_inches='tight',dpi=200);
```

```python
# 100k and 1M fill between plot for paper
prtf_min_100k = np.min(combs_PRTF, axis=0)
prtf_max_100k = np.max(combs_PRTF, axis=0)

prtf_min_1M = np.min(combs_PRTF_1M, axis=0)
prtf_max_1M = np.max(combs_PRTF_1M, axis=0)

plt.figure(dpi=140)
plt.xlim([0.0, res_PRTF_1M.max(axis=1)[0]])
plt.ylim([0.0, 1.02])

plt.xlabel('|q| $(nm^{-1})$', weight='bold')
plt.ylabel('PRTF', weight='bold')

plt.fill_between(res_ax, prtf_min_100k, prtf_max_100k, edgecolor="none", facecolor="b")
plt.fill_between(res_ax, prtf_min_1M, prtf_max_1M, edgecolor="none", facecolor="r")

plt.axhline(y=1/np.exp(1), xmin=0.0, xmax=res_ax[-1]*10, c='k', linestyle='--', linewidth=1.0, label='_nolegend_')
plt.axvline(x=1/0.6396483204611855, ymin=0 , ymax=1,c='k', linestyle='--', linewidth=1.0, alpha=0.4, label='_nolegend_')

plt.legend(["100k", "1M"], prop=dict(size=9, weight="bold"), frameon=False);

if saveFigs:
    plt.savefig(f'prtf_all_{extension}_filled.pdf', transparent=False, bbox_inches='tight',dpi=200);
```
