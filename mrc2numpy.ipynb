```python
import numpy as np
import scipy 
import scipy.fft as fft 
import scipy.ndimage as ndimage

import matplotlib.pyplot as plt

import h5py

import glob
import mrcfile

from helper_functions import write_text

saveDens = True

base_dir = 'average_reconstructions/chimerax/'
files_base_dir = sorted(glob.glob(base_dir + '*.mrc',recursive = True))
files_base_dir
```

<h2> Conversion of mrc file(s) </h2> 
After we have generated an electron density from a PDB structure in ChimeraX at a given resolution we save it as an mrc file. 
To work with the generated density, or for example a fitted map (using the fitting tools in ChimeraX) we need to
open and convert the mrc file to a NumPy array. We can also create our own mrc file if needed.

```python
mrc_name = files_base_dir
for f in mrc_name:
    with mrcfile.open(f, mode='r') as f_mod:
        dens = f_mod.data
        vox_size = f_mod.voxel_size
    print(f'Voxel size (x, y, z): {vox_size} Å')
    print(f'Array shape reshaping: {dens.shape}')

    if saveDens:
        write_text('Saving density!\n')
        np.save(f.split(sep='.mrc')[0]+'.npy', dens)
    else:
        write_text('Not saving density!\n')
```
