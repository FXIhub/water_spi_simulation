```python
import h5py
import numpy as np

import sys, glob

import matplotlib.pyplot as plt

global_dir = './'
```

<h2> Processing electron densities </h2> 

```python
base_dir = f'{global_dir}'
files = sorted(glob.glob(f'{base_dir}*.npy',recursive = True))
files
```

```python
for fc, f in enumerate(files):
    sys.stderr.write('\rConverting: %d/%d ...\n'%(fc + 1, len(files)))
    file = np.load(f)

    with h5py.File(f.split(sep='.npy')[0][:-5]+f'.h5', mode='w-') as conv:
        conv['density'] = file
```

One more step before R factor analysis is to convert the 3D Condor simulation into an h5 file with just the 3D intensities. 

```python
# Condor ground-truth intensity
gt_file =  '../debug_resampled_intensities/3d_groel_agipd_9_kev_denss.h5'

with h5py.File(gt_file, mode='r') as gt:
    gt_intens = gt['3Dparticle']['intensity'][:]
    
with h5py.File(gt_file.split(sep='.h5')[0]+'_trm.h5', mode='w-') as gt:
    gt['I'] = gt_intens
```
