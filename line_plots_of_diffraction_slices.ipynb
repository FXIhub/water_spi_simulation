```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

import seaborn as sns
sns.set_theme()
```

```python
slice_offset = 0

###### EMC diffraction pattern ######
emc_file = 'emc/protein_water_ds_4x_0001/data_100k_no_mask/prot_only_0/output_120.h5'

with h5py.File(emc_file, "r") as f:
    emc_diff = np.squeeze(f['intens'][:])

emc_diff = emc_diff[:-1,:-1,:-1]

center_emc = emc_diff.shape[0] // 2 + slice_offset
emc_diff_central_slice = emc_diff[center_emc][center_emc] ** (1/3)

###### Condor diffraction pattern ######
num_pats = 2000
def_rng = np.random.default_rng()
random_pat = def_rng.choice(num_pats)

sim_3d_file = '3d_groel_agipd_9_kev_denss_dsf_4x_375.h5'
with h5py.File(sim_3d_file, "r") as f:
    sim_3d_diff = f['3Dparticle/intensity'][:]

center_sim_3d = sim_3d_diff.shape[0] // 2 + slice_offset
sim_diff_central_slice_3d = sim_3d_diff[center_sim_3d][center_sim_3d] ** (1/3)

sim_file = f'sims_protein_water/run_{0}_protein_in_water_1000k_pats_dsf_4x/particle_intens.npy'
sim_diff = np.load(sim_file)[random_pat]
center_sim = sim_diff.shape[0] // 2 + slice_offset
sim_diff_central_slice = sim_diff[center_sim] ** (1/3)

##### Plotting of slices ######
fig_handle = plt.figure(constrained_layout = True, dpi = 300)
fig_handle.patch.set_facecolor(f'white')
spec_handle = fig_handle.add_gridspec(nrows = 1, ncols = 3)

ax_0 = fig_handle.add_subplot(spec_handle[0,0])
plt_0 = plt.plot(emc_diff_central_slice, 'bx--')
plt.xlabel('pixel #', weight='bold')
plt.xlim([center_emc-20, center_emc+20])
plt.ylim([0.0, None])
plt.title(f'EMC: {center_emc}/{emc_diff.shape[0]}', weight='bold');

ax_1 = fig_handle.add_subplot(spec_handle[0,1])
plt_1 = plt.plot(sim_diff_central_slice, 'rx--')
plt.xlabel('pixel #', weight='bold')
plt.xlim([center_sim-20, center_sim+20])
plt.ylim([0.0, None])
plt.title(f'2D Condor: {center_sim}/{sim_diff.shape[0]}', weight='bold');
    
ax_2 = fig_handle.add_subplot(spec_handle[0,2])
plt_2 = plt.plot(sim_diff_central_slice_3d, 'gx--')
plt.xlabel('pixel #', weight='bold')
plt.xlim([center_sim_3d-20, center_sim_3d+20])
plt.ylim([0.0, None])
plt.title(f'3D Condor: {center_sim_3d}/{sim_3d_diff.shape[0]}', weight='bold');
plt.savefig('4x_ds_diffraction_slices.pdf', transparent=False, bbox_inches='tight');
```
